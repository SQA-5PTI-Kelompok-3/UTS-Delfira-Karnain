```python
#ax^2+bx+c=0
```

```python
import math


def quadratic_solver(a, b, c):

    result = ""

    # Statement 1
    if a == 0:
        result = "Not Quadratic Equation"

    # Statement 2
    else:

        d = (b*b) - (4*a*c)


        # Decision 1
        if d > 0:

            x1 = (-b + math.sqrt(d))/(2*a)
            x2 = (-b - math.sqrt(d))/(2*a)

            result = (
                "Two roots : "
                + str(x1)
                + " and "
                + str(x2)
            )


        # Decision 2
        elif d == 0:

            x = -b/(2*a)

            result = (
                "One root : "
                + str(x)
            )


        else:

            result = "No real root"


    return result



# Input bebas user
a = float(input("Input a: "))
b = float(input("Input b: "))
c = float(input("Input c: "))


print(
    quadratic_solver(a,b,c)
)
```
